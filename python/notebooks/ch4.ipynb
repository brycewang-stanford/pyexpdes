This notebook reproduces the figures in `R files/ch4.R` using the Python implementation in `python/ch4.py`.

**Policy**
- Python-first implementation.
- If a specialized design generator is not practical to faithfully reproduce in Python, the Python code may call a **small R wrapper** to obtain the design points and then plot them in Python.
- **`minimaxdesign` is intentionally not used** due to package conflicts. Any code paths that would depend on `minimaxdesign` are omitted, and this is documented.

All figures are written to `python/figures/ch4/` and displayed below.


```python
import sys
from pathlib import Path
import importlib

python_dir = Path().resolve().parent
if str(python_dir) not in sys.path:
    sys.path.insert(0, str(python_dir))

import ch4
importlib.reload(ch4)

out_dir, fig_paths = ch4.generate_figures(use_r=True)
(out_dir, fig_paths)
```

```python
from IPython.display import Image, display
from pathlib import Path

missing = []
for p in fig_paths:
    pp = Path(p)
    if pp.exists():
        display(Image(filename=str(pp)))
    else:
        missing.append(str(pp))

if missing:
    print("Missing figure files (generation may have failed, or paths were stale):")
    print("\n".join(missing))
```
