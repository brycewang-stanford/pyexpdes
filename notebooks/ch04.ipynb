```python
import sys
from pathlib import Path
import importlib

src_dir = Path().resolve().parent / 'src'
if str(src_dir) not in sys.path:
    sys.path.insert(0, str(src_dir))

from pyexpdes.chapters import ch04_space_filling_designs
importlib.reload(ch04_space_filling_designs)

out_dir, fig_paths = ch04_space_filling_designs.generate_figures(use_r=True)
(out_dir, fig_paths)
```

```python
from IPython.display import Image, display
from pathlib import Path

missing = []
for p in fig_paths:
    pp = Path(p)
    if pp.exists():
        display(Image(filename=str(pp)))
    else:
        missing.append(str(pp))

if missing:
    print("Missing figure files (generation may have failed, or paths were stale):")
    print("\n".join(missing))
```
